<b><font size=5 color=pink >Step 6: integrate datasets</font></b>


```r
library(Seurat)
library(dplyr)

# =========================
# =========================

FBM_Stromal$dataset <- "FBM"
Adult_BM_Stromal$dataset <- "ABM"
Stromal$dataset <- "BMOs"
Organoids23_Stromal$dataset <- "BMO-2023"

DefaultAssay(FBM_Stromal) <- "RNA"
DefaultAssay(Adult_BM_Stromal) <- "RNA"
DefaultAssay(Stromal) <- "RNA"
DefaultAssay(Organoids23_Stromal) <- "RNA"


# =========================
# =========================

Stromal_merged <- merge(
  x = FBM_Stromal,
  y = list(
    Adult_BM_Stromal,
    Stromal,
    Organoids23_Stromal
  ),
  add.cell.ids = c(
    "FBM",
    "ABM",
    "BMOs",
    "BMO-2023"
  ),
  project = "Stromal_Integration"
)

table(Stromal_merged$dataset)

table(Stromal_merged$celltype)
```

```r
table(Stromal_merged$celltype, useNA = "ifany")
```

```r
Stromal_merged <- NormalizeData(Stromal_merged)
Stromal_merged <- FindVariableFeatures(Stromal_merged, selection.method = "vst", nfeatures = 2000)

all.genes <- rownames(Stromal_merged)
Stromal_merged <- ScaleData(Stromal_merged, features = all.genes)

Stromal_merged <- RunPCA(Stromal_merged, features = VariableFeatures(object = Stromal_merged))
```

```r
Stromal_merged <- IntegrateLayers(
  object = Stromal_merged,
  method = CCAIntegration,
  orig.reduction = "pca",
  new.reduction = "integrated.cca",
  verbose = TRUE
)

Stromal_merged[["RNA"]] <- JoinLayers(Stromal_merged[["RNA"]])
```

```r
DimHeatmap(Stromal_merged, dims = 1:30, cells = 500, balanced = TRUE)
ElbowPlot(Stromal_merged, ndims = 50)
```

```r
Stromal_merged <- FindNeighbors(Stromal_merged, reduction = "integrated.cca", dims = 1:30)
Stromal_merged <- FindClusters(Stromal_merged, resolution = 2)

Stromal_merged <- RunUMAP(Stromal_merged, reduction = "integrated.cca", dims = 1:30)
Stromal_merged <- RunTSNE(Stromal_merged, reduction = "integrated.cca", dims = 1:30)

p1 <- DimPlot(Stromal_merged, reduction = "tsne", label = TRUE) + ggtitle("t-SNE")
p2 <- DimPlot(Stromal_merged, reduction = "umap", label = TRUE) + ggtitle("UMAP")

p1+p2
```

<b><font size=5 color=pink >Step 8: stromal cell-type correlation</font></b>


```r
library(Seurat)
library(dplyr)
library(pheatmap)

# ============================================================
# ============================================================

DefaultAssay(Stromal_merged) <- "RNA"

Stromal_merged$celltype <- as.character(Stromal_merged$celltype)
Stromal_merged$dataset <- as.character(Stromal_merged$dataset)

Stromal_merged$group_plot <- as.character(Stromal_merged$group)

idx_na <- is.na(Stromal_merged$group_plot) |
  Stromal_merged$group_plot == ""

Stromal_merged$group_plot[idx_na] <-
  Stromal_merged$dataset[idx_na]

Stromal_merged$group_plot[
  Stromal_merged$group_plot %in% c("Adult_BM", "AdultBM", "Adult BM")
] <- "ABM"

Stromal_merged$group_plot[
  Stromal_merged$group_plot %in% c("Organoids23", "BMO2023", "BMO_2023")
] <- "BMO-2023"

table(Stromal_merged$group_plot, useNA = "ifany")
table(Stromal_merged$celltype, useNA = "ifany")
```

```r
# ============================================================
# ============================================================

keep_groups <- c(
  "ABM",
  "FBM",
  "Static.25d",
  "Dynamic.25d",
  "Dynamic.31d"
)

Stromal_sim <- subset(
  Stromal_merged,
  subset = group_plot %in% keep_groups &
    !is.na(celltype) &
    celltype != ""
)

Stromal_sim$group_celltype <- paste(
  Stromal_sim$group_plot,
  Stromal_sim$celltype,
  sep = " | "
)

table(Stromal_sim$group_plot, useNA = "ifany")
table(Stromal_sim$group_celltype)
```

```r
library(dplyr)
library(tidyr)

keep_groups <- c(
  "ABM",
  "FBM",
  "Static.25d",
  "Dynamic.25d",
  "Dynamic.31d"
)

cell_count_df <- Stromal_merged@meta.data %>%
  dplyr::mutate(
    group_plot = as.character(group_plot),
    celltype = as.character(celltype)
  ) %>%
  dplyr::filter(
    group_plot %in% keep_groups,
    !is.na(celltype),
    celltype != ""
  ) %>%
  dplyr::count(group_plot, celltype, name = "n_cells") %>%
  dplyr::arrange(group_plot, celltype)
```

```r
cell_count_wide <- cell_count_df %>%
  tidyr::pivot_wider(
    names_from = group_plot,
    values_from = n_cells,
    values_fill = 0
  )

cell_count_wide
```

```r
min_cells <- 30

keep_group_celltype <- cell_count_df %>%
  dplyr::filter(n_cells >= min_cells) %>%
  dplyr::mutate(group_celltype = paste(group_plot, celltype, sep = " | ")) %>%
  dplyr::pull(group_celltype)

Stromal_sim <- Stromal_merged

Stromal_sim$group_celltype <- paste(
  Stromal_sim$group_plot,
  Stromal_sim$celltype,
  sep = " | "
)

Stromal_sim <- subset(
  Stromal_sim,
  subset = group_plot %in% keep_groups &
    group_celltype %in% keep_group_celltype
)

table(Stromal_sim$group_celltype)
```

```r
# ============================================================
# ============================================================

av <- AverageExpression(
  Stromal_sim,
  group.by = "group_celltype",
  assays = "RNA",
  slot = "data"
)

av <- av[[1]]
av <- as.matrix(av)

dim(av)
head(av[, 1:min(5, ncol(av))])
```

```r
# ============================================================
# ============================================================

av <- av[
  rowSums(!is.finite(av)) == 0,
  ,
  drop = FALSE
]

av <- av[
  rowSums(av) > 0,
  ,
  drop = FALSE
]

gene_sd <- apply(av, 1, sd)

av_use <- av[
  !is.na(gene_sd) & gene_sd > 0,
  ,
  drop = FALSE
]

dim(av_use)

exp <- cor(
  x = av_use,
  y = NULL,
  method = "spearman",
  use = "pairwise.complete.obs"
)

dim(exp)
```
